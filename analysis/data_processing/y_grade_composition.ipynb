```python
# ---------- GRAFICO 4: UNIONE CLASSI 8A/8A+ E 8B/8B+ ----------
# Unisci 8A/8A+ e 8B/8B+
gradi_uniti = [
    "6B/6B+", "6C/6C+", "7A", "7A+", "7B/7B+", "7C", "7C+", "8A/8A+", "8B/8B+"
]
# Calcolo manuale delle potenze a metà per le classi unite
def potenza_mezzo(n1, n2, base=3.17):
    return base ** ((n1 + n2) / 2)

def ln_potenza_mezzo(n1, n2, base=3.17):
    return np.log(base ** ((n1 + n2) / 2))

indice_n = [
   4, 5, 6, 7, 8, 9, 10, 11.5, 13.5
]

difficolta_unite = [
    df[df["Grado_Fontainbleau"] == "6B/6B+"]["Difficolta_3.17^n"].values[0],
    df[df["Grado_Fontainbleau"] == "6C/6C+"]["Difficolta_3.17^n"].values[0],
    df[df["Grado_Fontainbleau"] == "7A"]["Difficolta_3.17^n"].values[0],
    df[df["Grado_Fontainbleau"] == "7A+"]["Difficolta_3.17^n"].values[0],
    df[df["Grado_Fontainbleau"] == "7B/7B+"]["Difficolta_3.17^n"].values[0],
    df[df["Grado_Fontainbleau"] == "7C"]["Difficolta_3.17^n"].values[0],
    df[df["Grado_Fontainbleau"] == "7C+"]["Difficolta_3.17^n"].values[0],
    potenza_mezzo(11, 12), # 8A/8A+
    potenza_mezzo(13, 14)  # 8B/8B+
]
ln_difficolta_unite = [np.log(v) for v in difficolta_unite]
```

```python
plt.figure(figsize=(10, 5))
plt.plot(gradi_uniti, difficolta_unite, marker='o', label='Difficolta', color='b')
plt.xticks(rotation=45)
plt.title("Difficolta e ln(Difficolta) per gradi da 6B+ a 8B+ (con classi unite)")
plt.xlabel("Grado Fontainbleau")
plt.ylabel("Valore")
plt.legend()
plt.tight_layout()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

plt.figure(figsize=(10, 5))
plt.plot(gradi_uniti, ln_difficolta_unite, marker='o', label='Difficolta', color='r')
plt.xticks(rotation=45)
plt.title("Difficolta e ln(Difficolta) per gradi da 6B+ a 8B+ (con classi unite)")
plt.xlabel("Grado Fontainbleau")
plt.ylabel("Valore")
plt.legend()
plt.tight_layout()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
